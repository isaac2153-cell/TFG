```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import tensorflow as tf
from tensorflow import keras
```

```python
model = keras.models.load_model(
    "efficientnet_sin_congelar_aug.h5",
    compile=False
)
print("Modelo cargado correctamente")
print(f"Input shape: {model.input_shape}")
print(f"Output shape: {model.output_shape}")
```

```python
img_path = "/root/tfg/CNN2_final/test/metastatic/HobI17-485538505474__HobI17-485538505474 [x=56000,y=16000,w=1000,h=1000].png"
```

```python
def get_img_array(img_path, target_size=(512, 512)):
    img = keras.utils.load_img(img_path, target_size=target_size)
    array = keras.utils.img_to_array(img)
    array = np.expand_dims(array, axis=0)
    return array

img_array = get_img_array(img_path)
print(f"Shape del array: {img_array.shape}")
```

```python
pred = model.predict(img_array, verbose=0)[0][0]

# IMPORTANTE: pred es P(no_metastatic) porque Keras asigna clase 0 = no_metastatic (alfabético)
# Por eso pred=0.0001 significa que es metastatic con confianza del 99.99%
clase_predicha = "no_metastatic" if pred > 0.5 else "metastatic"
confianza = pred if pred > 0.5 else 1 - pred

print(f"P(no_metastatic): {pred:.4f}")
print(f"P(metastatic): {1-pred:.4f}")
print(f"Clase predicha: {clase_predicha}")
print(f"Confianza: {confianza*100:.2f}%")
```

```python
plt.figure(figsize=(6, 6))
plt.imshow(img_array[0].astype("uint8"))
plt.axis("off")
plt.title(f"Original — {clase_predicha} ({confianza*100:.1f}%)")
plt.show()
```

```python
efficientnet_base = model.get_layer("efficientnetb3")

last_conv_layer_model = keras.Model(
    inputs=efficientnet_base.inputs,
    outputs=efficientnet_base.get_layer("top_conv").output
)
print("Modelo extractor de top_conv creado")
print(f"Output shape: {last_conv_layer_model.output_shape}")
```

```python
classifier_input = keras.Input(shape=(16, 16, 1536))
x = model.get_layer("global_average_pooling2d_3")(classifier_input)
x = model.get_layer("dense_6")(x)
x = model.get_layer("dropout_3")(x)
classifier_output = model.get_layer("dense_7")(x)
classifier_model = keras.Model(classifier_input, classifier_output)

print("Modelo cabeza de clasificación creado")
print(f"Input shape: {classifier_model.input_shape}")
print(f"Output shape: {classifier_model.output_shape}")
```

```python
seq_layer = model.get_layer("sequential")
img_array_processed = seq_layer(img_array, training=False)

print(f"Shape original: {img_array.shape}")
print(f"Shape tras sequential: {img_array_processed.shape}")
```

```python
with tf.GradientTape() as tape:
    last_conv_output = last_conv_layer_model(img_array_processed)
    tape.watch(last_conv_output)
    
    preds = classifier_model(last_conv_output)
    
    # CORRECCIÓN: preds[:, 0] es P(no_metastatic) = 0.0001
    # Como la imagen ES metastatic, calculamos gradientes de la clase metastatic:
    loss = 1 - preds[:, 0]

grads = tape.gradient(loss, last_conv_output)

print(f"Shape de las activaciones: {last_conv_output.shape}")
print(f"Shape de los gradientes: {grads.shape}")
print(f"P(no_metastatic): {preds.numpy()[0][0]:.4f}")
print(f"Score usado para Grad-CAM (P metastatic): {loss.numpy()[0]:.4f}")
```

```python
pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2)).numpy()

last_conv_output_np = last_conv_output.numpy()[0]
for i in range(pooled_grads.shape[-1]):
    last_conv_output_np[:, :, i] *= pooled_grads[i]

heatmap = np.mean(last_conv_output_np, axis=-1)

# ReLU: solo nos interesa lo que activa positivamente
heatmap = np.maximum(heatmap, 0)

# CORRECCIÓN: evitar división por cero cuando el heatmap es todo negativo o cero
if np.max(heatmap) > 0:
    heatmap /= np.max(heatmap)
else:
    print("AVISO: heatmap todo a cero. Revisa que estás calculando gradientes de la clase correcta.")

print(f"Shape del heatmap: {heatmap.shape}")
print(f"Valor mínimo: {heatmap.min():.4f}")
print(f"Valor máximo: {heatmap.max():.4f}")

plt.figure(figsize=(6, 6))
plt.imshow(heatmap, cmap="jet")
plt.colorbar()
plt.title("Mapa de calor Grad-CAM (16x16) — Clase: metastatic")
plt.axis("off")
plt.show()
```

```python
# Usar el colormap jet sin deprecation warning
jet = plt.colormaps["jet"]

# heatmap ya está normalizado entre 0 y 1 (de la celda [11])
heatmap_uint8 = np.uint8(255 * heatmap)
jet_heatmap = jet(heatmap_uint8)[:, :, :3]  # Tomar solo RGB, descartar canal alpha

print(f"Shape del heatmap RGB: {jet_heatmap.shape}")
```

```python
# Cargar la imagen ORIGINAL (1000x1000, tamaño nativo del tile)
img_original = keras.utils.load_img(img_path)
img_original = keras.utils.img_to_array(img_original)

# Redimensionar heatmap de 16x16 a 1000x1000
jet_heatmap_img = keras.utils.array_to_img(jet_heatmap)
jet_heatmap_resized = jet_heatmap_img.resize((img_original.shape[1], img_original.shape[0]))
jet_heatmap_array = keras.utils.img_to_array(jet_heatmap_resized)

print(f"Shape de la imagen original: {img_original.shape}")
print(f"Shape del heatmap redimensionado: {jet_heatmap_array.shape}")
```

```python
alpha = 0.4

# CORRECCIÓN: la fórmula era img * (1 - alpha * 0.5), lo cual está mal
# Debe ser: heatmap * alpha + imagen_original * (1 - alpha)
superimposed = jet_heatmap_array * alpha + img_original * (1 - alpha)
superimposed = np.clip(superimposed, 0, 255)
superimposed_img = keras.utils.array_to_img(superimposed)

plt.figure(figsize=(8, 8))
plt.imshow(superimposed_img)
plt.title(f"{clase_predicha} ({confianza*100:.1f}%)")
plt.axis("off")
plt.savefig("gradcam_resultado4.png", dpi=150, bbox_inches="tight")
plt.show()
```
