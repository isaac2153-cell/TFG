```python
import pathlib
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# CONFIGURACIÓN DE GPU
gpus = tf.config.experimental.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
        print("GPU disponible y configurada correctamente.")
    except RuntimeError as e:
        print("Error al configurar GPU:", e)
else:
    print("No se detectó GPU.")

# CONFIGURACIÓN DE MATPLOTLIB
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.grid"] = False

print("Num GPUs disponibles:", len(tf.config.list_physical_devices('GPU')))
print("TensorFlow version:", tf.__version__)
```

```python
train_ds = keras.utils.image_dataset_from_directory(
    "/workspace/Desktop/CNN1/Train",
    image_size=(512, 512),
    batch_size=32,
    label_mode='categorical'
)

val_ds = keras.utils.image_dataset_from_directory(
    "/workspace/Desktop/CNN1/Val",
    image_size=(512, 512),
    batch_size=32,
    label_mode='categorical',
    shuffle=False
)

test_ds = keras.utils.image_dataset_from_directory(
    "/workspace/Desktop/CNN1/Test",
    image_size=(512, 512),
    batch_size=32,
    label_mode='categorical',
    shuffle=False
)
```

```python
conv_base  = keras.applications.EfficientNetB3(
    weights="imagenet",
    include_top=False,
    input_shape=(512, 512, 3)
)

conv_base.trainable = False
```

```python
inputs = keras.Input(shape=(512, 512, 3))

x = conv_base(inputs)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(256, activation="relu")(x)
x = layers.Dropout(0.5)(x)
outputs = layers.Dense(3, activation="softmax")(x)

model = keras.Model(inputs, outputs)

model.compile(loss="categorical_crossentropy",
              optimizer=keras.optimizers.Adam(learning_rate=1e-4),
              metrics=["accuracy"])
```

```python
callbacks = [
    keras.callbacks.ModelCheckpoint(
        filepath="Multi_efficientnet_congelado.h5",
        save_best_only=True,
        monitor="val_loss"),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=3,
        min_lr=1e-6)
]
history = model.fit(
    train_ds,
    epochs=50,
    validation_data=val_ds,
    callbacks=callbacks)
```

```python
test_model = keras.models.load_model("Multi_efficientnet_congelado.h5")
test_loss, test_acc = test_model.evaluate(test_ds)
print(f"Test accuracy: {test_acc:.3f}")
```

```python
accuracy = history.history["accuracy"]
val_accuracy = history.history["val_accuracy"]
loss = history.history["loss"]
val_loss = history.history["val_loss"]
epochs = range(1, len(accuracy) + 1)
plt.plot(epochs, accuracy, "bo", label="Training accuracy")
plt.plot(epochs, val_accuracy, "b", label="Validation accuracy")
plt.title("Training and validation accuracy")
plt.legend()
plt.figure()
plt.plot(epochs, loss, "bo", label="Training loss")
plt.plot(epochs, val_loss, "b", label="Validation loss")
plt.title("Training and validation loss")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (
    confusion_matrix, classification_report,
    roc_curve, auc, precision_recall_curve,
    f1_score, precision_score, recall_score,
    accuracy_score, matthews_corrcoef
)

# Cargar el modelo
test_model = keras.models.load_model("Multi_efficientnet_congelado.h5")

# Predicciones: ahora devuelve (n_samples, 3)
y_pred_proba = test_model.predict(test_ds)

# Etiquetas reales: vienen en one-hot (n_samples, 3)
y_true_onehot = np.concatenate([y for x, y in test_ds], axis=0)

# Convertimos a enteros (0, 1, 2)
y_pred = np.argmax(y_pred_proba, axis=1)
y_true = np.argmax(y_true_onehot, axis=1)

print(f"Total test samples: {len(y_true)}")
print(f"Reales sano: {(y_true == 0).sum()}")
print(f"Reales lobulillar: {(y_true == 1).sum()}")
print(f"Reales ductal: {(y_true == 2).sum()}")
print(f"Predichos sano: {(y_pred == 0).sum()}")
print(f"Predichos lobulillar: {(y_pred == 1).sum()}")
print(f"Predichos ductal: {(y_pred == 2).sum()}")
```

```python
from sklearn.preprocessing import label_binarize

# Cargar modelo y predecir
test_model = keras.models.load_model("Multi_efficientnet_congelado.h5")
y_pred_proba = test_model.predict(test_ds)
y_true_onehot = np.concatenate([y for x, y in test_ds], axis=0)

y_pred = np.argmax(y_pred_proba, axis=1)
y_true = np.argmax(y_true_onehot, axis=1)

class_names = test_ds.class_names

# Métricas generales
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, average='macro', zero_division=0)
recall = recall_score(y_true, y_pred, average='macro', zero_division=0)
f1 = f1_score(y_true, y_pred, average='macro', zero_division=0)
mcc = matthews_corrcoef(y_true, y_pred)

# Matriz de confusión
cm = confusion_matrix(y_true, y_pred)

# Specificity y NPV por clase (One-vs-Rest)
specificities = []
npvs = []
for i in range(3):
    tp = cm[i, i]
    fp = cm[:, i].sum() - tp
    fn = cm[i, :].sum() - tp
    tn = cm.sum() - tp - fp - fn
    spec = tn / (tn + fp) if (tn + fp) > 0 else 0.0
    npv = tn / (tn + fn) if (tn + fn) > 0 else 0.0
    specificities.append(spec)
    npvs.append(npv)

specificity_macro = np.mean(specificities)
npv_macro = np.mean(npvs)

# AUC-ROC y AUC-PR (One-vs-Rest)
y_true_bin = label_binarize(y_true, classes=[0, 1, 2])

roc_aucs = []
pr_aucs = []
for i in range(3):
    fpr, tpr, _ = roc_curve(y_true_bin[:, i], y_pred_proba[:, i])
    roc_aucs.append(auc(fpr, tpr))
    prec, rec, _ = precision_recall_curve(y_true_bin[:, i], y_pred_proba[:, i])
    pr_aucs.append(auc(rec, prec))

roc_auc_macro = np.mean(roc_aucs)
pr_auc_macro = np.mean(pr_aucs)

# Print final
print("="*55)
print("RESULTADOS - EfficientNet 512×512 (Multiclase)")
print("="*55)
print(f"Accuracy:            {accuracy:.4f}")
print(f"Precision (macro):   {precision:.4f}")
print(f"Recall/Sens (macro): {recall:.4f}")
print(f"Specificity (macro): {specificity_macro:.4f}")
print(f"NPV (macro):         {npv_macro:.4f}")
print(f"F1-score (macro):    {f1:.4f}")
print(f"MCC:                 {mcc:.4f}")
print(f"AUC-ROC (macro OvR): {roc_auc_macro:.4f}")
print(f"AUC-PR (macro OvR):  {pr_auc_macro:.4f}")
print("="*55)
print(f"\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=class_names))
print(f"\nMatriz de confusión:")
print(f"{'':>12}  Pred sano  Pred lob  Pred duct")
for i, name in enumerate(class_names):
    print(f"Real {name:>10}  {cm[i,0]:>9}  {cm[i,1]:>8}  {cm[i,2]:>9}")
```

```python
class_names = test_ds.class_names

# --- Matriz de confusión ---
cm = confusion_matrix(y_true, y_pred)

# --- ROC y PR en One-vs-Rest ---
y_true_bin = label_binarize(y_true, classes=[0, 1, 2])

# Calcular curvas por clase
roc_data = []
pr_data = []
for i, name in enumerate(class_names):
    # ROC
    fpr, tpr, _ = roc_curve(y_true_bin[:, i], y_pred_proba[:, i])
    roc_auc_i = auc(fpr, tpr)
    roc_data.append((fpr, tpr, roc_auc_i, name))
    
    # PR
    prec, rec, _ = precision_recall_curve(y_true_bin[:, i], y_pred_proba[:, i])
    pr_auc_i = auc(rec, prec)
    pr_data.append((rec, prec, pr_auc_i, name))

# --- Gráficas ---
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Matriz de confusión
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0],
            xticklabels=class_names, yticklabels=class_names)
axes[0].set_xlabel('Predicción')
axes[0].set_ylabel('Real')
axes[0].set_title('Matriz de Confusión')

# 2. Curva ROC (One-vs-Rest)
colors = ['darkblue', 'darkorange', 'darkgreen']
for i, (fpr, tpr, roc_auc_i, name) in enumerate(roc_data):
    axes[1].plot(fpr, tpr, color=colors[i], linewidth=2,
                 label=f'{name} (AUC = {roc_auc_i:.4f})')
axes[1].plot([0, 1], [0, 1], 'k--', linewidth=1)
axes[1].set_xlabel('1 - Especificidad (FPR)')
axes[1].set_ylabel('Sensibilidad (TPR)')
axes[1].set_title('Curva ROC - One vs Rest')
axes[1].legend()
axes[1].grid(alpha=0.3)

# 3. Curva Precision-Recall (One-vs-Rest)
for i, (rec, prec, pr_auc_i, name) in enumerate(pr_data):
    axes[2].plot(rec, prec, color=colors[i], linewidth=2,
                 label=f'{name} (AUC = {pr_auc_i:.4f})')
axes[2].set_xlabel('Recall')
axes[2].set_ylabel('Precision')
axes[2].set_title('Curva Precision-Recall - One vs Rest')
axes[2].legend()
axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.savefig("Multi_efficientnet_congelado.png", dpi=150)
plt.show()
```
