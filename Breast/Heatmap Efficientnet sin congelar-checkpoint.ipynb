**Configuración incial**

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import tensorflow as tf
from tensorflow import keras
```

*Cargar modelo entrenado*

```python
model = keras.models.load_model(
    "Efficientnet_breast_sin_congelar.h5",
    compile=False
)
print("Modelo cargado correctamente")
print(f"Input shape: {model.input_shape}")
print(f"Output shape: {model.output_shape}")
```

**Preparar la imagen**

```python
img_path = "/workspace/Desktop/CNN1/Test/Invasive_carcinoma/TCGA-D8-A27T-01Z-00-DX1.1E3A4D57-9CF2-4EBF-B74D-ADD7BD8CBFA5 [x=13000,y=30000,w=1000,h=1000].png"
```

*Función de carga y preprocesdado*

```python
def get_img_array(img_path, target_size=(512, 512)):
    img = keras.utils.load_img(img_path, target_size=target_size)
    array = keras.utils.img_to_array(img)
    array = np.expand_dims(array, axis=0)
    return array

img_array = get_img_array(img_path)
print(f"Shape del array: {img_array.shape}")
```

*Predicción del modelo sobre la imagen*

```python
pred = model.predict(img_array, verbose=0)[0][0]
clase_predicha = "Carcinoma" if pred > 0.5 else "Sano"
confianza = pred if pred > 0.5 else 1 - pred

print(f"Probabilidad predicha: {pred:.4f}")
print(f"Clase predicha: {clase_predicha}")
print(f"Confianza: {confianza*100:.2f}%")
```

*Ver la imagen original*

```python
plt.figure(figsize=(6, 6))
plt.imshow(img_array[0].astype("uint8"))
plt.axis("off")
plt.title(f"Original — {clase_predicha} ({confianza*100:.1f}%)")
plt.show()
```

**Crear mapa de calor Grad-CAM**

*Modelo que extrae última capa convolucional*

```python
efficientnet_base = model.get_layer("efficientnetb3")

last_conv_layer_model = keras.Model(
    inputs=efficientnet_base.inputs,
    outputs=efficientnet_base.get_layer("top_activation").output
)
print("Modelo extractor de top_activation creado")
print(f"Output shape: {last_conv_layer_model.output_shape}")
```

*Reconstruir la cabeza de clasificación*

```python
classifier_input = keras.Input(shape=(16, 16, 1536))
x = model.get_layer("global_average_pooling2d")(classifier_input)
x = model.get_layer("dense")(x)
x = model.get_layer("dropout")(x)
classifier_output = model.get_layer("dense_1")(x)
classifier_model = keras.Model(classifier_input, classifier_output)

print("Modelo cabeza de clasificación creado")
print(f"Input shape: {classifier_model.input_shape}")
print(f"Output shape: {classifier_model.output_shape}")
```

*Cálculo de gradientes con GradientTape*

```python
with tf.GradientTape() as tape:
    # 1. Pasamos la imagen por la base de EfficientNet
    last_conv_output = last_conv_layer_model(img_array)
    tape.watch(last_conv_output)
    
    # 2. Pasamos las activaciones por la cabeza
    preds = classifier_model(last_conv_output)
    
    # 3. Como es binario con sigmoid, usamos directamente la salida
    loss = preds[:, 0]

# Gradientes de la predicción respecto a las activaciones convolucionales
grads = tape.gradient(loss, last_conv_output)

print(f"Shape de las activaciones: {last_conv_output.shape}")
print(f"Shape de los gradientes: {grads.shape}")
```

*Construir heatmap a partir de los gradientes*

```python
# Promediar los gradientes sobre las dimensiones espaciales
pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2)).numpy()

# Multiplicar cada canal de activación por su peso correspondiente
last_conv_output = last_conv_output.numpy()[0]
for i in range(pooled_grads.shape[-1]):
    last_conv_output[:, :, i] *= pooled_grads[i]

# Promediar los canales para obtener el heatmap final
heatmap = np.mean(last_conv_output, axis=-1)

# Normalizar entre 0 y 1
heatmap = np.maximum(heatmap, 0)
heatmap /= np.max(heatmap)

print(f"Shape del heatmap: {heatmap.shape}")
print(f"Valor mínimo: {heatmap.min():.4f}")
print(f"Valor máximo: {heatmap.max():.4f}")
```

*Ver heatmap en crudo*

```python
plt.figure(figsize=(6, 6))
plt.matshow(heatmap, cmap="jet", fignum=1)
plt.colorbar()
plt.title("Mapa de calor Grad-CAM (16x16)")
plt.axis("off")
plt.show()
```

**Superposición del heatmap sobre la imagen original**

*Convertir heatmpa a imagen RGB con colormap jet*

```python
heatmap_uint8 = np.uint8(255 * heatmap)
jet = cm.get_cmap("jet")
jet_colors = jet(np.arange(256))[:, :3]
jet_heatmap = jet_colors[heatmap_uint8]

print(f"Shape del heatmap RGB: {jet_heatmap.shape}")
```

*Redimensionar el heatmap al tamaño de la imagen original*

```python
img = keras.utils.load_img(img_path)
img = keras.utils.img_to_array(img)

jet_heatmap = keras.utils.array_to_img(jet_heatmap)
jet_heatmap = jet_heatmap.resize((img.shape[1], img.shape[0]))
jet_heatmap = keras.utils.img_to_array(jet_heatmap)

print(f"Shape de la imagen original: {img.shape}")
print(f"Shape del heatmap redimensionado: {jet_heatmap.shape}")
```

*Superposición y visualización final*

```python
alpha = 0.4 

superimposed_img = jet_heatmap * alpha + img * (1 - alpha * 0.5)
superimposed_img = np.clip(superimposed_img, 0, 255)
superimposed_img = keras.utils.array_to_img(superimposed_img)

plt.figure(figsize=(8, 8))
plt.imshow(superimposed_img)
plt.title(f"{clase_predicha} ({confianza*100:.1f}%)")
plt.axis("off")
plt.savefig("gradcam_resultado4.png", dpi=150, bbox_inches="tight")
plt.show()
```
